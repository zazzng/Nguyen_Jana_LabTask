### Part 3: Filtering ###

1. Show only the rows where the age is above 60.

2. Show rows where the fare is above 100 and the passenger was in 1st class.

3. Identify and isolate rows with missing values in the age column.

```python
# 1. Show only the rows where the age is above 60
tdf = df[df['Age'] > 60]
print('1. Show only the rows where the age is above 60', tdf)

# 2. Show rows where the fare is above 100 and the passenger was in 1st class
abc = df[(df['Fare'] > 100) & (df['Pclass'] == 1)]
print('2. Show rows where the fare is above 100 and the passenger was in 1st class', abc)

# 3. Identify and isolate rows with missing values in the age column
mar = df.loc[df['Age'].isna()]
print('3. Identify and isolate rows with missing values in the age column', mar)
```

### Part 4: Reindexing and Dropping ###

1. Reverse the row order of the dataset.

2. Remove one column from the dataset.

3. Remove all rows with missing values in the age column.

```python
# 1. Reverse the row order of the dataset
dataReversed = df[::-1]
print('1. Reverse the row order of the dataset', dataReversed)

# 2. Remove one column from the dataset
removedColumn = df.drop(columns=['Name'])
print('2. Remove one column from the dataset', removedColumn)

# 3. Remove all rows with missing values in the age column
missingAge = df.dropna(subset=['Age'])
print('3. Remove all rows with missing values in the age column', missingAge)
```

### Part 5: Arithmetic and Data Alignment ###

1. Create a new column that converts fares from USD to Euros.

2. Subtract the average age from every age value.

3. Subtract a custom array from two selected columns in the dataset.

```python
# 1. Create a new column that converts fares from USD to Euros
df['Fare_EUR'] = df['Fare'] * 0.92
print('1. Create a new column that converts fares from USD to Euros', df[['Fare', 'Fare_EUR']])

# 2. Subtract the average age from every age value
# Average
averageAge = df['Age'].mean()
# Subtract
df['Subtracted'] = df['Age'] - averageAge
print('2. Subtract the average age from every age value', df[['Age', 'Subtracted']])

# 3. Subtract a custom array from two selected columns in the dataset
df_fixed = df[['Fare', 'Age']] - [10, 2]
print('3. Subtract a custom array from two selected columns in the dataset', df_fixed)
```

### Part 7: Mapping and Transformation ###
1. Convert categorical values in the gender column to numeric codes.

2. Create a new column that categorizes age into groups (e.g., child, adult, senior).



```python
# 1. Convert categorical values in the gender column to numeric codes
df['SexNum'] = df['Sex'].map({'male': 0, 'female': 1})
print(df[['Sex', 'SexNum']])

# 2. Create a new column that categorizes age into groups (e.g., child, adult, senior)
conditions = [
    df['Age'] <= 12,
    (df['Age'] > 12) & (df['Age'] <= 18),
    (df['Age'] > 18) & (df['Age'] <= 60),
    df['Age'] > 60
]

choices = ['Child', 'Teen', 'Adult', 'Senior']

df['Age_Group'] = np.select(conditions, choices, default='Unknown')

print('2. Create a new column that categorizes age into groups (e.g., child, adult, senior)', df[['Age', 'Age_Group']]
```

```python
# PART 2 - 1
# 1. Select the names column
import csv
import urllib.request
url = "https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv"
response = urllib.request.urlopen(url)
lines = [line.decode("utf-8") for line in response.readlines()]

reader = csv.DictReader(lines)
names = [row["Name"] for row in reader]

print('1. Select the names column', names[:10])

#PART 3 - 1
# 1. Filter rows where Age > 60
print('3-1. Filter rows where Age > 60')
reader = csv.DictReader(lines)
older_than_60 = []

for row in reader:
    age = row["Age"]
    
    if age == "" or age is None:
        continue
    
    if float(age) > 60:
        older_than_60.append(row)

for row in older_than_60:
    print(row["Name"], row["Age"])

# PART 5 - 1
# 1. Create a new column that converts fares from USD to Euros
print('5-1. Create a new column that converts fares from USD to Euros')
USD_TO_EUR = 0.92

reader = csv.DictReader(lines)
updated_rows = []

for row in reader:
    fare = row["Fare"]

    if fare == "" or fare is None:
        row["Fare_EUR"] = None
    else:
        row["Fare_EUR"] = float(fare) * USD_TO_EUR

    updated_rows.append(row)

for row in updated_rows[:10]:
    print(row["Name"], row["Fare"], row["Fare_EUR"])
```
